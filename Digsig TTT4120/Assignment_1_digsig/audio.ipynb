Problem 2b + plotting for sanity check

```python
import numpy as np
import matplotlib.pyplot as plt

def harmonic_sequence(A, f_1, n_array):
    return A*np.cos(2*np.pi*f_1*n_array)

#F_S = 6000Hz -> 24k samples in 4 sec 
F_S = 6000
F_1 = 2000
f_1 = F_1/F_S

n_values = np.arange(24000)
more_n_values = np.linspace(0,4,1000000)
x_n = harmonic_sequence(1,f_1, n_values)
cos_values = np.cos(2*np.pi*2000*more_n_values)

plt.xlim(0, 0.002)
plt.plot(more_n_values, cos_values, label="2000 Hz cosine")
plt.scatter(n_values/6000, x_n, color="red", s=5, label="Samples")
plt.show()
```

Problem 2c

```python
import numpy as np
import matplotlib.pyplot as plt
import sounddevice as sd

def harmonic_sequence(A, f_1, n_array):
    return A*np.cos(2*np.pi*f_1*n_array)

F_S1, F_S2, F_S3 = 1000, 3000, 12000
f_1 = 0.3
#Number of samples in 4 sec
N_1, N_2, N_3 = 4*F_S1, 4*F_S2, 4*F_S3

n_1_values = np.arange(N_1)
n_2_values = np.arange(N_2)
n_3_values = np.arange(N_3)

x_n1 = harmonic_sequence(1,f_1, n_1_values)
x_n2 = harmonic_sequence(1,f_1, n_2_values)
x_n3 = harmonic_sequence(1,f_1, n_3_values)

#Pitch increases because F_1 increases
sd.play(x_n1, samplerate=F_S1)
sd.wait()
sd.play(x_n2, samplerate=F_S2)
sd.wait()
sd.play(x_n3, samplerate=F_S3)
sd.wait()
```

Problem 2d

```python
import numpy as np
import matplotlib.pyplot as plt
import sounddevice as sd

def harmonic_sequence(A, f_1, n_array):
    return A*np.cos(2*np.pi*f_1*n_array)

F_S = 8000
F1_1000, F1_3000, F1_6000 = 1000, 3000, 6000

#Number of samples in 4 sec
N = 4*F_S
f1_0, f1_1, f1_2 = F1_1000/F_S, F1_3000/F_S, F1_6000/F_S

n_values = np.arange(N)

x_n1 = harmonic_sequence(1,f1_0, n_values)
x_n2 = harmonic_sequence(1,f1_1, n_values)
x_n3 = harmonic_sequence(1,f1_2, n_values)


#2000Hz note, f1 = 0.125 (f1 < 0.5)
sd.play(x_n1, samplerate=F_S)
sd.wait()
#3000Hz note f1 = 0.375 (f1 < 0.5)
sd.play(x_n2, samplerate=F_S)
sd.wait()
#2000Hz note because of aliasing (F1 was 6000Hz). f1 = 0.75 and f > 0.5
sd.play(x_n3, samplerate=F_S)
sd.wait()
```
