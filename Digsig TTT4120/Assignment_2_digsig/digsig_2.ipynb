```python
import numpy as np
import matplotlib.pyplot as plt

M = 10
omega = np.linspace(-np.pi, np.pi, 2000)
Y = np.sin((M + 0.5) * omega) / np.sin(omega / 2)

plt.figure(figsize=(8,4))
plt.plot(omega, Y, 'b')
plt.title("DTFT of y[n], M=10")
plt.xlabel(r"$\omega$")
plt.ylabel(r"$Y(\omega)$")
plt.grid(True)
plt.show()
```
